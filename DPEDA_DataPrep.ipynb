```python
#Number of Shades in international brand
plt.figure(figsize=(10,6))

brand_counts = inter['brand'].value_counts().reset_index()
brand_counts.columns = ['brand', 'shade_count']


heatmap_data = brand_counts.set_index('brand').T

sns.heatmap(heatmap_data,
           annot=True,
           fmt='d',
           cmap='YlOrRd',
           cbar_kws={'label': 'Number of Shades'},
           linewidths=0.5)

plt.title("Total Shade Count per Brand - Heatmap")
plt.xlabel("Brand")
plt.ylabel("")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Combining indian shade ranges into one DataFrame
indian = pd.concat([may_ind, mac_ind,loreal_india], ignore_index=True)
```

```python
#Undertone Distribution by Brand of Indian Shades

#Create pie charts for each brand
brands = indian['brand'].unique()
fig, axes = plt.subplots(1, len(brands), figsize=(5*len(brands), 5))

for i, brand in enumerate(brands):
    data = indian[indian['brand'] == brand]['undertone'].value_counts()
    axes[i].pie(data.values, labels=data.index, autopct='%1.1f%%',
                colors=sns.color_palette("Set3", len(data)))
    axes[i].set_title(brand)

plt.suptitle("Undertone Distribution by Brand")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7,5))
brand_counts = indian["brand"].value_counts()
sns.barplot(x=brand_counts.index, y=brand_counts.values, palette="Set3")
plt.title("Total Shade Count per Brand")
plt.xlabel("Brand")
plt.ylabel("Number of Shades")
plt.show()
```

```python
#undertone distribution by brands in India(heatmap)
#cross-tabulation of brand and undertone
undertone_brand_counts = pd.crosstab(indian['brand'], indian['undertone'])

# Create the heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(undertone_brand_counts, annot=True, fmt='d', cmap='YlGnBu')
plt.title('Undertone Distribution by Brand (Indian Shades)')
plt.xlabel('Undertone')
plt.ylabel('Brand')
plt.show()
```

```python
#undertone distribution by brands internationally(heatmap)
#cross-tabulation of brand and undertone for international shades
undertone_brand_counts_international = pd.crosstab(inter['brand'], inter['undertone'])

# Create the heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(undertone_brand_counts_international, annot=True, fmt='d', cmap='YlGnBu')
plt.title('Undertone Distribution by Brand (International Shades)')
plt.xlabel('Undertone')
plt.ylabel('Brand')
plt.show()
```

```python
#total shade count India vs Internatinal
indian['brand'] = indian['brand'].str.strip().str.title()
inter['brand'] = inter['brand'].str.strip().str.title()

indian_shade_counts = indian['brand'].value_counts().reset_index()
indian_shade_counts.columns = ['brand', 'indian_shade_count']

international_shade_counts = inter['brand'].value_counts().reset_index()
international_shade_counts.columns = ['brand', 'international_shade_count']

merged_shade_counts = pd.merge(
    indian_shade_counts,
    international_shade_counts,
    on='brand',
    how='outer'
).fillna(0)

plt.figure(figsize=(8,6))
merged_shade_counts.set_index('brand').plot(
    kind='bar',
    stacked=False,
    color=['#1f77b4', '#ff7f0e'],
    width=0.7
)

plt.title("Total Shade Count per Brand (Indian vs International)", fontsize=14, fontweight="bold")
plt.xlabel("Brand", fontsize=12)
plt.ylabel("Number of Shades", fontsize=12)
plt.xticks(rotation=0)
plt.legend(title="Dataset", labels=["Indian Shades", "International Shades"])
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()
```
